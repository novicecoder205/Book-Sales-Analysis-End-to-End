```python
plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

##### Histogram Result:

- Groups publishing years into intervals (bins).
 
- Shows the total number of books within each interval.
 
- The 2000–2020 range has the highest frequency because many books were published during those years.

```python
df["Publishing Year"].value_counts().sort_index().plot(
    kind="line",
    figsize=(14, 5),
    marker="o",
    title="Book Publishing Trends Over Time"
)
```

##### Line Chart Result:

- Shows the number of books published in each individual year.

- You can see publishing activity increasing from 2000 onwards.

- The sharp decline near 2020 means fewer books were published in those final years.
