```python
plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

##### Histogram Result:

- Groups publishing years into intervals (bins).
 
- Shows the total number of books within each interval.
 
- The 2000–2020 range has the highest frequency because many books were published during those years.

```python
df["Publishing Year"].value_counts().sort_index().plot(
    kind="line",
    figsize=(14, 5),
    marker="o",
    title="Book Publishing Trends Over Time"
)
```

##### Line Chart Result:

- Shows the number of books published in each individual year.

- You can see publishing activity increasing from 2000 onwards.

- The sharp decline near 2020 means fewer books were published in those final years.

```python
sns.boxplot(
    x="genre",
    y="Book_ratings_count",
    data=df,
    hue="genre",
    palette="Set2",
    legend=False
)
plt.xlabel("Genre")
plt.ylabel("Book Ratings Count") # also mean total No. of ratings received by an individual book in each category.
plt.title("Box Plot of Book ratings Count for Each Genre")
plt.show()
```

- Genre Fiction: Has the highest median book ratings count (~92,000), suggesting that books in this genre generally receive more ratings.
  
- Fiction & Nonfiction: Have similar median ratings counts (~82,000).
  
- Children: Has the lowest median (~72,000), suggesting fewer ratings compared to the other genres.
  
- Box (Middle 50%): Genre fiction has a wider distribution, while children's books have a narrower distribution.

  
- Outliers: All genres except children have several unusually high-rating books, with genre fiction reaching over 200,000 ratings.

  
Overall conclusion: Genre fiction tends to receive the most ratings, while children's books receive the fewest. However, individual books can still achieve exceptionally high ratings counts regardless of genre.

```python
plt.scatter(df["sale price"], df["units sold"])
plt.xlabel("Sale Price")
plt.ylabel("Units Sold")
plt.title("Scatter Plot of Sale Price vs Units Sold")
plt.show()
```

- X-axis (Sale Price): Represents the selling price of each book.
  
- Y-axis (Units Sold): Represents the number of copies sold.
  
- Data Distribution: Most books are concentrated between $1–$13 in sale price.
  
- High Sales: Some books have exceptionally high sales, reaching approximately 60,000 units.
  
- Low Sales: A large number of books have relatively low sales, generally below 7,000 units.
  
- Relationship: There doesn't appear to be a clear positive or negative correlation between sale price and units sold.
  
Overall conclusion: Book prices do not appear to have a strong relationship with the number of units sold. Some relatively inexpensive books achieve high sales, but higher prices do not necessarily mean lower sales.
